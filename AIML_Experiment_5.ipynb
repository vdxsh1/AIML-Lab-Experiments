```python
import math
print("MATH LIBRARY")
print("Square Root:", math.sqrt(25))
print("Factorial:", math.factorial(5))
print("Power:", math.pow(2, 3))
print("Value of Pi:", math.pi)
import numpy as np
print("\nNUMPY LIBRARY")
arr = np.array([10, 20, 30, 40, 50])
print("Array:", arr)
print("Sum:", np.sum(arr))
print("Mean:", np.mean(arr))
print("Maximum:", np.max(arr))
print("Minimum:", np.min(arr))
from scipy import stats
print("\nSCIPY LIBRARY")
data = [10, 20, 20, 30, 40, 50]
print("Mean:", stats.tmean(data))
print("Median:", np.median(data))
print("Mode:", stats.mode(data))
print("Standard Deviation:", np.std(data))
import pandas as pd
print("\nPANDAS LIBRARY")
data = {
    'Name': ['Amit', 'Riya', 'John', 'Sara'],
    'Marks': [85, 90, 75, 95]
}
df = pd.DataFrame(data)
print("DataFrame:")
print(df)
print("Average Marks:", df['Marks'].mean())
import matplotlib.pyplot as plt
print("\nMATPLOTLIB LIBRARY")
plt.plot(df['Name'], df['Marks'], marker='o')
plt.title("Student Marks")
plt.xlabel("Students")
plt.ylabel("Marks")
plt.show()
import seaborn as sns
print("\nSEABORN LIBRARY")
sns.barplot(x='Name', y='Marks', data=df)
plt.title("Student Marks Comparison")
plt.show()
```
